```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/customer_cleaned.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["age"],
    bins=15,
    kde=True
)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["income"],
    bins=20,
    kde=True
)

plt.title("Customer Income Distribution")
plt.xlabel("Income")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["annual_spending"],
    bins=20,
    kde=True
)

plt.title("Annual Customer Spending Distribution")
plt.xlabel("Annual Spending")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["purchase_frequency"],
    bins=10,
    kde=True
)

plt.title("Purchase Frequency Distribution")
plt.xlabel("Purchase Frequency")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x="gender"
)

plt.title("Customer Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="education"
)

plt.title("Customer Distribution by Education")
plt.xlabel("Education")
plt.ylabel("Number of Customers")
plt.xticks(rotation=20)
plt.show()
```

```python
df[[
    "age",
    "income",
    "purchase_frequency",
    "annual_spending"
]].describe().T
```

```python
correlation_matrix = df[
    [
        "age",
        "income",
        "purchase_frequency",
        "annual_spending"
    ]
].corr()

correlation_matrix
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Customer Feature Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="purchase_frequency",
    y="annual_spending"
)

plt.title("Purchase Frequency vs Annual Spending")
plt.xlabel("Purchase Frequency")
plt.ylabel("Annual Spending")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="income",
    y="annual_spending"
)

plt.title("Income vs Annual Spending")
plt.xlabel("Income")
plt.ylabel("Annual Spending")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="age",
    y="annual_spending"
)

plt.title("Age vs Annual Spending")
plt.xlabel("Age")
plt.ylabel("Annual Spending")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="gender",
    y="annual_spending"
)

plt.title("Annual Spending by Gender")
plt.xlabel("Gender")
plt.ylabel("Annual Spending")
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="education",
    y="annual_spending"
)

plt.title("Annual Spending by Education Level")
plt.xlabel("Education")
plt.ylabel("Annual Spending")
plt.xticks(rotation=20)
plt.show()
```
